```python
model = "claude-opus-5"
```

```python
from anthropic import AnthropicFoundry
```

```python
client = AnthropicFoundry(
    api_key="",
    base_url=""
)
```

```python
message = client.messages.create(
    model=model, 
    system="Support agent",
    messages=[{
        "role": "user",
        "content": "Tell me about Ai foundry in 3 sentence"
    }],
    max_tokens=1024
)
```
